## OK onwards with our first project

````python
# A class to represent a Webpage
# If you're not familiar with Classes, check out the "Intermediate Python" notebook

# Some websites need you to use proper headers when fetching them:
headers = {
 "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}

class Website:

    def __init__(self, url):
        """
        Create this Website object from the given url using the BeautifulSoup library
        """
        self.url = url
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.content, 'html.parser')
        self.title = soup.title.string if soup.title else "No title found"
        for irrelevant in soup.body(["script", "style", "img", "input"]):
            irrelevant.decompose()
        self.text = soup.body.get_text(separator="\n", strip=True)
````

````python
# Let's try one out. Change the website and add print statements to follow along.

ed = Website("https://edwarddonner.com")
print(ed.title)
print(ed.text)
````

## Types of prompts

You may know this already - but if not, you will get very familiar with it!

Models like GPT4o have been trained to receive instructions in a particular way.

They expect to receive:

**A system prompt** that tells them what task they are performing and what tone they should use

**A user prompt** -- the conversation starter that they should reply to

````python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = "You are an assistant that analyzes the contents of a website \
and provides a short summary, ignoring text that might be navigation related. \
Respond in markdown."
````

````python
# A function that writes a User Prompt that asks for summaries of websites:

def user_prompt_for(website):
    user_prompt = f"You are looking at a website titled {website.title}"
    user_prompt += "\nThe contents of this website is as follows; \
please provide a short summary of this website in markdown. \
If it includes news or announcements, then summarize these too.\n\n"
    user_prompt += website.text
    return user_prompt
````

````python
print(user_prompt_for(ed))
````

## Messages

The API from OpenAI expects to receive messages in a particular structure.
Many of the other APIs share this structure:

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```
To give you a preview, the next 2 cells make a rather simple call - we won't stretch the mighty GPT (yet!)

````python
messages = [
    {"role": "system", "content": "You are a snarky assistant"},
    {"role": "user", "content": "What is 2 + 2?"}
]
````

````python
# To give you a preview -- calling OpenAI with system and user messages:

response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
print(response.choices[0].message.content)
````

## And now let's build useful messages for GPT-4o-mini, using a function

````python
# See how this function creates exactly the format above

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_for(website)}
    ]
````

## Time to bring it together - the API for OpenAI is very simple!

````python
# And now: call the OpenAI API. You will get very familiar with this!

def summarize(url):
    website = Website(url)
    response = openai.chat.completions.create(
        model = "gpt-4o-mini",
        messages = messages_for(website)
    )
    return response.choices[0].message.content
````

````python
summarize("https://edwarddonner.com")
````

````python
# A function to display this nicely in the Jupyter output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
````

## Solution: Enhanced Website Class with Playwright

Here's an improved Website class that can handle JavaScript-rendered websites like OpenAI's site using Playwright. This solution will work with modern websites that rely on JavaScript to render content.

First, you'll need to install Playwright:

````python
# Install playwright (run this once)
!pip install playwright
!playwright install chromium
````

````python
# Enhanced Website class that handles JavaScript-rendered content
import subprocess
import tempfile
import json
import os
import sys

class EnhancedWebsite:
    def __init__(self, url, use_playwright=False, wait_time=2):
        """
        Create this Website object from the given url.
        
        Args:
            url: The URL to scrape
            use_playwright: If True, use Playwright for JavaScript-rendered content
            wait_time: Time to wait for JavaScript to load (in seconds)
        """
        self.url = url
        
        if use_playwright:
            self._scrape_with_playwright(wait_time)
        else:
            self._scrape_with_requests()
    
    def _scrape_with_requests(self):
        """Original scraping method using requests"""
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
        }
        try:
            response = requests.get(self.url, headers=headers, timeout=10)
            soup = BeautifulSoup(response.content, 'html.parser')
            self.title = soup.title.string if soup.title else "No title found"
            
            # Remove irrelevant elements
            if soup.body:
                for irrelevant in soup.body(["script", "style", "img", "input"]):
                    irrelevant.decompose()
                self.text = soup.body.get_text(separator="\n", strip=True)
            else:
                self.text = "No body content found"
        except Exception as e:
            self.title = f"Error: {str(e)}"
            self.text = f"Could not fetch content: {str(e)}"
    
    def _scrape_with_playwright(self, wait_time):
        """Enhanced scraping method using Playwright in a subprocess"""
        
        # Create a temporary script to run playwright
        script_content = f'''
import asyncio
import sys
import traceback
import json

try:
    from playwright.async_api import async_playwright
except ImportError:
    result = {{"title": "Import Error", "text": "Playwright not installed. Please run: pip install playwright", "success": False}}
    print(json.dumps(result, indent=2))
    sys.exit(0)

async def scrape():
    try:
        async with async_playwright() as p:
            try:
                # Launch browser with more realistic settings
                browser = await p.chromium.launch(
                    headless=True,
                    args=[
                        '--no-sandbox',
                        '--disable-blink-features=AutomationControlled',
                        '--disable-features=VizDisplayCompositor'
                    ]
                )
                
                # Create context with realistic user agent
                context = await browser.new_context(
                    user_agent='Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36',
                    viewport={{'width': 1280, 'height': 720}}
                )
                
                page = await context.new_page()
                
                # Remove webdriver property
                await page.add_init_script("""
                    Object.defineProperty(navigator, 'webdriver', {{
                        get: () => undefined,
                    }});
                """)
                
                try:
                    # Navigate to the page with better error handling
                    response = await page.goto("{self.url}", wait_until='domcontentloaded', timeout=30000)
                    
                    if response is None:
                        raise Exception("Failed to load page - no response")
                    
                    if response.status >= 400:
                        raise Exception(f"HTTP error {{response.status}}: {{response.status_text}}")
                    
                    # Wait for JavaScript to load
                    await asyncio.sleep({wait_time})
                    
                    # Get the title with better error handling
                    try:
                        title = await page.title()
                        if not title or title.strip() == '':
                            # Try alternative methods to get title
                            title_alternatives = await page.evaluate("""
                                () => {{
                                    return document.querySelector('title')?.textContent || 
                                           document.querySelector('h1')?.textContent || 
                                           document.querySelector('[role="heading"]')?.textContent ||
                                           'No title found';
                                }}
                            """)
                            title = title_alternatives or "No title found"
                    except Exception as title_error:
                        title = f"Title extraction error: {{str(title_error)}}"
                    
                    # Get text content with multiple robust methods
                    text = "No content found"
                    
                    try:
                        # Wait a bit more for dynamic content
                        await page.wait_for_load_state('networkidle', timeout=10000)
                    except:
                        pass  # Continue even if network idle times out
                    
                    try:
                        # Try multiple methods to extract text
                        text = await page.evaluate("""
                            () => {{
                                try {{
                                    // Remove unwanted elements
                                    const unwanted = document.querySelectorAll('script, style, nav, header, footer, .advertisement, .ad, [class*="ad"], [id*="ad"]');
                                    unwanted.forEach(el => {{
                                        try {{ el.remove(); }} catch(e) {{}}
                                    }});
                                    
                                    // Get main content with priority order
                                    let content = '';
                                    
                                    // Try main content areas first
                                    const contentSelectors = [
                                        'main', '[role="main"]', '.content', '#content', 
                                        'article', '.article', '.post', '.entry-content',
                                        '.main-content', '.page-content', '.container'
                                    ];
                                    
                                    for (const selector of contentSelectors) {{
                                        const element = document.querySelector(selector);
                                        if (element && element.innerText && element.innerText.length > 100) {{
                                            content = element.innerText;
                                            break;
                                        }}
                                    }}
                                    
                                    // Fallback to body if no main content found
                                    if (!content) {{
                                        content = document.body?.innerText || document.body?.textContent || '';
                                    }}
                                    
                                    // Clean up the content
                                    if (content) {{
                                        // Remove excessive whitespace and clean up
                                        content = content.replace(/\\s+/g, ' ').trim();
                                        // Remove common unwanted text
                                        content = content.replace(/^(.*?)?(Skip to content|Skip to main content)/i, '');
                                        content = content.replace(/(Cookie|Privacy Policy|Terms of Service).*$/i, '');
                                    }}
                                    
                                    return content || 'No readable content found';
                                    
                                }} catch (e) {{
                                    return `Text extraction error: ${{e.toString()}}`;
                                }}
                            }}
                        """)
                        
                    except Exception as eval_error:
                        # Ultimate fallback: try to get any visible text
                        try:
                            text = await page.inner_text('body')
                            if text == "Unknown Error":
                                # This might be a blocked page, try one more method
                                text = await page.content()
                                if len(text) > 1000:  # If we got HTML content
                                    text = "Page loaded but text extraction was blocked by the website"
                                else:
                                    text = f"Text extraction failed: {{str(eval_error)}}"
                        except:
                            text = f"All text extraction methods failed: {{str(eval_error)}}"
                    
                    # Final text processing
                    if isinstance(text, str) and text:
                        # Limit length and clean up
                        if len(text) > 10000:
                            text = text[:10000] + "... [truncated]"
                        
                        # Check for common blocking messages
                        if any(blocked_msg in text.lower() for blocked_msg in ['unknown error', 'access denied', 'blocked', 'forbidden']):
                            text = f"Website appears to be blocking automated access. Original message: {{text[:200]}}"
                    else:
                        text = "No readable content could be extracted"
                    
                    result = {{"title": title, "text": text, "success": True}}
                    
                except Exception as page_error:
                    result = {{"title": "Page loading error", "text": f"Could not access page: {{str(page_error)}}", "success": False}}
                finally:
                    await browser.close()
                    
            except Exception as browser_error:
                result = {{"title": "Browser error", "text": f"Could not launch browser: {{str(browser_error)}}", "success": False}}
                
    except Exception as playwright_error:
        result = {{"title": "Playwright error", "text": f"Could not initialize Playwright: {{str(playwright_error)}}", "success": False}}
    
    print(json.dumps(result, indent=2))

if __name__ == "__main__":
    try:
        asyncio.run(scrape())
    except Exception as e:
        error_result = {{"title": "Runtime error", "text": f"Failed to run scraper: {{str(e)}}", "success": False}}
        print(json.dumps(error_result, indent=2))
'''
        
        try:
            # Create temporary script file
            with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as f:
                f.write(script_content)
                temp_script = f.name
            
            # Get the current Python executable to ensure we use the same environment
            python_executable = sys.executable
            
            # Run the script in a subprocess
            result = subprocess.run([python_executable, temp_script], 
                                  capture_output=True, text=True, timeout=60)
            
            if result.returncode == 0 and result.stdout.strip():
                try:
                    data = json.loads(result.stdout.strip())
                    self.title = data.get('title', 'No title found')
                    self.text = data.get('text', 'No content found')
                    
                    # Only show debug info if there's actually an error
                    if not data.get('success', False):
                        print(f"Playwright reported an error: {data}")
                        
                except json.JSONDecodeError as je:
                    self.title = "JSON parsing error"
                    self.text = f"Could not parse result. Raw output: {result.stdout[:200]}..."
            else:
                self.title = "Subprocess error"
                if result.stderr:
                    self.text = f"STDERR: {result.stderr[:500]}"
                else:
                    self.text = f"No valid output from subprocess. Return code: {result.returncode}"
                
        except subprocess.TimeoutExpired:
            self.title = "Timeout error"
            self.text = "The scraping operation timed out after 60 seconds"
        except Exception as e:
            self.title = f"Unexpected error: {str(e)}"
            self.text = f"Could not scrape with Playwright: {str(e)}"
        finally:
            # Clean up temporary file
            try:
                os.unlink(temp_script)
            except:
                pass
````

````python
# Enhanced summarization functions that can handle JavaScript content

def enhanced_user_prompt_for(website):
    """Create user prompt for the enhanced website object"""
    user_prompt = f"You are looking at a website titled {website.title}"
    user_prompt += "\nThe contents of this website is as follows; \
please provide a short summary of this website in markdown. \
If it includes news or announcements, then summarize these too.\n\n"
    user_prompt += website.text
    return user_prompt

def enhanced_messages_for(website):
    """Create messages for the enhanced website object"""
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": enhanced_user_prompt_for(website)}
    ]

def enhanced_summarize(url, use_playwright=False, wait_time=2):
    """
    Enhanced summarize function that can handle JavaScript content
    
    Args:
        url: URL to summarize
        use_playwright: If True, use Playwright for JavaScript content
        wait_time: Time to wait for JavaScript to load
    """
    website = EnhancedWebsite(url, use_playwright=use_playwright, wait_time=wait_time)
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=enhanced_messages_for(website)
    )
    return response.choices[0].message.content

def enhanced_display_summary(url, use_playwright=False, wait_time=2):
    """Display enhanced summary with markdown formatting"""
    summary = enhanced_summarize(url, use_playwright=use_playwright, wait_time=wait_time)
    display(Markdown(summary))
````

## Key Differences

1. **Regular requests method**: Fast, lightweight, but can't execute JavaScript
2. **Playwright method**: Slower, requires more resources, but handles JavaScript-rendered content

## When to Use Each Method

- Use **requests** (original method) for:
  - Simple, static websites
  - Speed is critical
  - You don't need JavaScript execution

- Use **Playwright** for:
  - Modern web applications (React, Vue, Angular)
  - Sites that load content dynamically
  - When requests method returns incomplete content

The enhanced class gives you the flexibility to choose the right tool for each website!
