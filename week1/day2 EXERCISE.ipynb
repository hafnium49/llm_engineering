## Are you confused about why that works?

It seems strange, right? We just used OpenAI code to call Ollama?? What's going on?!

Here's the scoop:

The python class `OpenAI` is simply code written by OpenAI engineers that makes calls over the internet to an endpoint.  

When you call `openai.chat.completions.create()`, this python code just makes a web request to the following url: "https://api.openai.com/v1/chat/completions"

Code like this is known as a "client library" - it's just wrapper code that runs on your machine to make web requests. The actual power of GPT is running on OpenAI's cloud behind this API, not on your computer!

OpenAI was so popular, that lots of other AI providers provided identical web endpoints, so you could use the same approach.

So Ollama has an endpoint running on your local box at http://localhost:11434/v1/chat/completions  
And in week 2 we'll discover that lots of other providers do this too, including Gemini and DeepSeek.

And then the team at OpenAI had a great idea: they can extend their client library so you can specify a different 'base url', and use their library to call any compatible API.

That's it!

So when you say: `ollama_via_openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')`  
Then this will make the same endpoint calls, but to Ollama instead of OpenAI.

Take the code from day1 and incorporate it here, to build a website summarizer that uses Llama 3.2 running locally instead of OpenAI; use either of the above approaches.

```python
# SOLUTION: Website Summarizer with Local Llama 3.2
# This combines the enhanced website scraping from Day 1 with Ollama

# First, let's create the basic Website class from Day 1
class Website:
    def __init__(self, url):
        """
        Create this Website object from the given url using the BeautifulSoup library
        """
        self.url = url
        response = requests.get(url, headers={'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'}, timeout=10)
        soup = BeautifulSoup(response.content, 'html.parser')
        self.title = soup.title.string if soup.title else "No title found"
        
        if soup.body:
            for irrelevant in soup.body(["script", "style", "img", "input"]):
                irrelevant.decompose()
            self.text = soup.body.get_text(separator="\n", strip=True)
        else:
            self.text = "No body content found"

# Create the summarization functions
system_prompt = "You are an assistant that analyzes the contents of a website and provides a short summary, ignoring text that might be navigation related. Respond in markdown."

def user_prompt_for(website):
    user_prompt = f"You are looking at a website titled {website.title}"
    user_prompt += "\nThe contents of this website is as follows; please provide a short summary of this website in markdown. If it includes news or announcements, then summarize these too.\n\n"
    user_prompt += website.text
    return user_prompt

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_for(website)}
    ]

# Website summarizer using Ollama (Method 1: Direct HTTP API)
def summarize_with_http_api(url):
    """Summarize website using direct HTTP API calls to Ollama"""
    print(f"🌐 Scraping website: {url}")
    website = Website(url)
    print(f"📖 Title: {website.title}")
    print(f"📄 Content length: {len(website.text)} characters")
    
    payload = {
        "model": MODEL,
        "messages": messages_for(website),
        "stream": False
    }
    
    try:
        print("🤖 Generating summary with Llama 3.2 (HTTP API)...")
        response = requests.post(OLLAMA_API, json=payload, headers=HEADERS)
        response.raise_for_status()
        summary = response.json()['message']['content']
        print("✅ Summary generated successfully!")
        return summary
    except Exception as e:
        return f"❌ Error generating summary: {str(e)}"

# Website summarizer using Ollama (Method 2: ollama package)
def summarize_with_ollama_package(url):
    """Summarize website using the ollama Python package"""
    import ollama
    
    print(f"🌐 Scraping website: {url}")
    website = Website(url)
    print(f"📖 Title: {website.title}")
    print(f"📄 Content length: {len(website.text)} characters")
    
    try:
        print("🤖 Generating summary with Llama 3.2 (ollama package)...")
        response = ollama.chat(model=MODEL, messages=messages_for(website))
        summary = response['message']['content']
        print("✅ Summary generated successfully!")
        return summary
    except Exception as e:
        return f"❌ Error generating summary: {str(e)}"

# Website summarizer using Ollama (Method 3: OpenAI client format)
def summarize_with_openai_format(url):
    """Summarize website using OpenAI client format pointing to Ollama"""
    print(f"🌐 Scraping website: {url}")
    website = Website(url)
    print(f"📖 Title: {website.title}")
    print(f"📄 Content length: {len(website.text)} characters")
    
    try:
        # Use OpenAI client format pointing to Ollama
        ollama_client = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
        
        print("🤖 Generating summary with Llama 3.2 (OpenAI format)...")
        response = ollama_client.chat.completions.create(
            model=MODEL,
            messages=messages_for(website)
        )
        
        summary = response.choices[0].message.content
        print("✅ Summary generated successfully!")
        return summary
    except Exception as e:
        return f"❌ Error generating summary: {str(e)}"

# Display function with nice formatting
def display_summary(url, method="openai_format"):
    """Display website summary with nice markdown formatting"""
    print("\n" + "="*60)
    print(f"SUMMARIZING: {url}")
    print("="*60)
    
    if method == "http_api":
        summary = summarize_with_http_api(url)
    elif method == "ollama_package":
        summary = summarize_with_ollama_package(url)
    else:  # default to openai_format
        summary = summarize_with_openai_format(url)
    
    print("\n📝 SUMMARY:")
    print("-" * 30)
    display(Markdown(summary))
    print("="*60)

print("✅ Website summarizer functions loaded successfully!")
print("💡 Use display_summary('your-url-here') to test!")
```

## BONUS: Enhanced Website Scraper (From Day 1)

For JavaScript-heavy websites like OpenAI.com, we can use the enhanced scraper with Playwright support:

```python
# Enhanced Website class that can handle JavaScript-rendered content
# This is the improved version from Day 1 that works with sites like OpenAI.com

import subprocess
import tempfile
import json
import os

class EnhancedWebsite:
    def __init__(self, url, use_playwright=False, wait_time=2):
        """
        Create this Website object from the given url.
        
        Args:
            url: The URL to scrape
            use_playwright: If True, use Playwright for JavaScript-rendered content
            wait_time: Time to wait for JavaScript to load (in seconds)
        """
        self.url = url
        
        if use_playwright:
            self._scrape_with_playwright(wait_time)
        else:
            self._scrape_with_requests()
    
    def _scrape_with_requests(self):
        """Original scraping method using requests"""
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
        }
        try:
            response = requests.get(self.url, headers=headers, timeout=10)
            soup = BeautifulSoup(response.content, 'html.parser')
            self.title = soup.title.string if soup.title else "No title found"
            
            if soup.body:
                for irrelevant in soup.body(["script", "style", "img", "input"]):
                    irrelevant.decompose()
                self.text = soup.body.get_text(separator="\n", strip=True)
            else:
                self.text = "No body content found"
        except Exception as e:
            self.title = f"Error: {str(e)}"
            self.text = f"Could not fetch content: {str(e)}"
    
    def _scrape_with_playwright(self, wait_time):
        """Enhanced scraping method using Playwright in a subprocess"""
        
        script_content = f'''
import asyncio
import sys
import json

try:
    from playwright.async_api import async_playwright
except ImportError:
    result = {{"title": "Import Error", "text": "Playwright not installed. Run: pip install playwright", "success": False}}
    print(json.dumps(result, indent=2))
    sys.exit(0)

async def scrape():
    try:
        async with async_playwright() as p:
            browser = await p.chromium.launch(
                headless=True,
                args=['--no-sandbox', '--disable-blink-features=AutomationControlled']
            )
            
            context = await browser.new_context(
                user_agent='Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36',
                viewport={{'width': 1280, 'height': 720}}
            )
            
            page = await context.new_page()
            
            try:
                response = await page.goto("{self.url}", wait_until='domcontentloaded', timeout=30000)
                await asyncio.sleep({wait_time})
                
                title = await page.title() or "No title found"
                
                text = await page.evaluate("""
                    () => {{
                        const unwanted = document.querySelectorAll('script, style, nav, header, footer, .advertisement, .ad');
                        unwanted.forEach(el => {{ try {{ el.remove(); }} catch(e) {{}} }});
                        
                        return document.body?.innerText || document.body?.textContent || 'No content found';
                    }}
                """)
                
                if len(text) > 10000:
                    text = text[:10000] + "... [truncated]"
                
                result = {{"title": title, "text": text, "success": True}}
                
            except Exception as page_error:
                result = {{"title": "Page error", "text": f"Could not load page: {{str(page_error)}}", "success": False}}
            finally:
                await browser.close()
                
    except Exception as e:
        result = {{"title": "Error", "text": f"Playwright error: {{str(e)}}", "success": False}}
    
    print(json.dumps(result, indent=2))

if __name__ == "__main__":
    asyncio.run(scrape())
'''
        
        try:
            with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as f:
                f.write(script_content)
                temp_script = f.name
            
            result = subprocess.run([sys.executable, temp_script], 
                                  capture_output=True, text=True, timeout=60)
            
            if result.returncode == 0 and result.stdout.strip():
                data = json.loads(result.stdout.strip())
                self.title = data.get('title', 'No title found')
                self.text = data.get('text', 'No content found')
            else:
                self.title = "Subprocess error"
                self.text = f"Could not scrape with Playwright: {result.stderr}"
                
        except Exception as e:
            self.title = f"Error: {str(e)}"
            self.text = f"Could not scrape with Playwright: {str(e)}"
        finally:
            try:
                os.unlink(temp_script)
            except:
                pass

# Enhanced summarizer that can handle JavaScript sites
def enhanced_summarize_with_ollama(url, use_playwright=False):
    """Enhanced summarizer using EnhancedWebsite and local Llama 3.2"""
    print(f"🌐 Scraping website: {url} {'(with Playwright)' if use_playwright else '(basic)'}")
    
    website = EnhancedWebsite(url, use_playwright=use_playwright)
    print(f"📖 Title: {website.title}")
    print(f"📄 Content length: {len(website.text)} characters")
    
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_for(website)}
    ]
    
    try:
        ollama_client = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
        
        print("🤖 Generating summary with Llama 3.2...")
        response = ollama_client.chat.completions.create(
            model=MODEL,
            messages=messages
        )
        
        summary = response.choices[0].message.content
        print("✅ Summary generated successfully!")
        return summary
    except Exception as e:
        return f"❌ Error generating summary: {str(e)}"

def enhanced_display_summary(url, use_playwright=False):
    """Display enhanced summary with nice formatting"""
    print("\n" + "="*60)
    print(f"ENHANCED SUMMARIZING: {url}")
    if use_playwright:
        print("🎭 Using Playwright for JavaScript content")
    print("="*60)
    
    summary = enhanced_summarize_with_ollama(url, use_playwright)
    
    print("\n📝 SUMMARY:")
    print("-" * 30)
    display(Markdown(summary))
    print("="*60)

print("✅ Enhanced website scraper loaded!")
print("💡 Use enhanced_display_summary('url', use_playwright=True) for JS sites!")
```

# Website Summarizer with Local Llama 3.2

Now let's combine the enhanced website scraping capabilities from Day 1 with our local Llama 3.2 setup to create a powerful website summarizer that runs entirely locally!

This implementation will:
1. Use both BeautifulSoup (for simple sites) and Playwright (for JavaScript-heavy sites)
2. Automatically detect when to use which scraping method
3. Use our local Llama 3.2 model instead of OpenAI
4. Handle errors gracefully and provide fallback options

```python
# Enhanced Website class from Day 1 with Playwright support
import subprocess
import tempfile
import json
import sys

# Some websites need proper headers when fetching them:
headers = {
 "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}

class EnhancedWebsite:
    def __init__(self, url, use_playwright=False, wait_time=2):
        """
        Create this Website object from the given url.
        
        Args:
            url: The URL to scrape
            use_playwright: If True, use Playwright for JavaScript-rendered content
            wait_time: Time to wait for JavaScript to load (in seconds)
        """
        self.url = url
        
        if use_playwright:
            self._scrape_with_playwright(wait_time)
        else:
            self._scrape_with_requests()
    
    def _scrape_with_requests(self):
        """Original scraping method using requests + BeautifulSoup"""
        try:
            response = requests.get(self.url, headers=headers, timeout=10)
            soup = BeautifulSoup(response.content, 'html.parser')
            self.title = soup.title.string if soup.title else "No title found"
            
            # Remove irrelevant elements
            if soup.body:
                for irrelevant in soup.body(["script", "style", "img", "input"]):
                    irrelevant.decompose()
                self.text = soup.body.get_text(separator="\n", strip=True)
            else:
                self.text = "No body content found"
        except Exception as e:
            self.title = f"Error: {str(e)}"
            self.text = f"Could not fetch content: {str(e)}"
    
    def _scrape_with_playwright(self, wait_time):
        """Enhanced scraping method using Playwright in a subprocess"""
        
        # Create a temporary script to run playwright
        script_content = f'''
import asyncio
import sys
import json

try:
    from playwright.async_api import async_playwright
except ImportError:
    result = {{"title": "Import Error", "text": "Playwright not installed. Please run: pip install playwright", "success": False}}
    print(json.dumps(result, indent=2))
    sys.exit(0)

async def scrape():
    try:
        async with async_playwright() as p:
            try:
                # Launch browser with realistic settings
                browser = await p.chromium.launch(
                    headless=True,
                    args=[
                        '--no-sandbox',
                        '--disable-blink-features=AutomationControlled',
                        '--disable-features=VizDisplayCompositor'
                    ]
                )
                
                # Create context with realistic user agent
                context = await browser.new_context(
                    user_agent='Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36',
                    viewport={{'width': 1280, 'height': 720}}
                )
                
                page = await context.new_page()
                
                # Remove webdriver property to avoid detection
                await page.add_init_script("""
                    Object.defineProperty(navigator, 'webdriver', {{
                        get: () => undefined,
                    }});
                """)
                
                try:
                    # Navigate to the page
                    response = await page.goto("{self.url}", wait_until='domcontentloaded', timeout=30000)
                    
                    if response is None or response.status >= 400:
                        raise Exception(f"HTTP error {{response.status if response else 'No response'}}")
                    
                    # Wait for JavaScript to load
                    await asyncio.sleep({wait_time})
                    
                    # Get the title
                    try:
                        title = await page.title() or "No title found"
                    except:
                        title = "Title extraction failed"
                    
                    # Get text content with multiple robust methods
                    try:
                        # Wait a bit more for dynamic content
                        await page.wait_for_load_state('networkidle', timeout=10000)
                    except:
                        pass  # Continue even if network idle times out
                    
                    try:
                        # Extract main content
                        text = await page.evaluate("""
                            () => {{
                                try {{
                                    // Remove unwanted elements
                                    const unwanted = document.querySelectorAll('script, style, nav, header, footer, .advertisement, .ad');
                                    unwanted.forEach(el => {{
                                        try {{ el.remove(); }} catch(e) {{}}
                                    }});
                                    
                                    // Get main content with priority selectors
                                    const contentSelectors = [
                                        'main', '[role="main"]', '.content', '#content', 
                                        'article', '.article', '.post', '.entry-content',
                                        '.main-content', '.page-content', '.container'
                                    ];
                                    
                                    let content = '';
                                    for (const selector of contentSelectors) {{
                                        const element = document.querySelector(selector);
                                        if (element && element.innerText && element.innerText.length > 100) {{
                                            content = element.innerText;
                                            break;
                                        }}
                                    }}
                                    
                                    // Fallback to body
                                    if (!content) {{
                                        content = document.body?.innerText || document.body?.textContent || '';
                                    }}
                                    
                                    // Clean up content
                                    if (content) {{
                                        content = content.replace(/\\s+/g, ' ').trim();
                                    }}
                                    
                                    return content || 'No readable content found';
                                    
                                }} catch (e) {{
                                    return `Text extraction error: ${{e.toString()}}`;
                                }}
                            }}
                        """)
                        
                    except Exception as eval_error:
                        try:
                            text = await page.inner_text('body')
                        except:
                            text = f"All text extraction methods failed: {{str(eval_error)}}"
                    
                    # Final text processing
                    if isinstance(text, str) and text:
                        if len(text) > 10000:
                            text = text[:10000] + "... [truncated]"
                    else:
                        text = "No readable content could be extracted"
                    
                    result = {{"title": title, "text": text, "success": True}}
                    
                except Exception as page_error:
                    result = {{"title": "Page loading error", "text": f"Could not access page: {{str(page_error)}}", "success": False}}
                finally:
                    await browser.close()
                    
            except Exception as browser_error:
                result = {{"title": "Browser error", "text": f"Could not launch browser: {{str(browser_error)}}", "success": False}}
                
    except Exception as playwright_error:
        result = {{"title": "Playwright error", "text": f"Could not initialize Playwright: {{str(playwright_error)}}", "success": False}}
    
    print(json.dumps(result, indent=2))

if __name__ == "__main__":
    try:
        asyncio.run(scrape())
    except Exception as e:
        error_result = {{"title": "Runtime error", "text": f"Failed to run scraper: {{str(e)}}", "success": False}}
        print(json.dumps(error_result, indent=2))
'''
        
        try:
            # Create temporary script file
            with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as f:
                f.write(script_content)
                temp_script = f.name
            
            # Run the script in a subprocess
            result = subprocess.run([sys.executable, temp_script], 
                                  capture_output=True, text=True, timeout=60)
            
            if result.returncode == 0 and result.stdout.strip():
                try:
                    data = json.loads(result.stdout.strip())
                    self.title = data.get('title', 'No title found')
                    self.text = data.get('text', 'No content found')
                    
                    if not data.get('success', False):
                        print(f"Playwright warning: {data}")
                        
                except json.JSONDecodeError:
                    self.title = "JSON parsing error"
                    self.text = f"Could not parse result. Raw output: {result.stdout[:200]}..."
            else:
                self.title = "Subprocess error"
                if result.stderr:
                    self.text = f"STDERR: {result.stderr[:500]}"
                else:
                    self.text = f"No valid output from subprocess. Return code: {result.returncode}"
                
        except subprocess.TimeoutExpired:
            self.title = "Timeout error"
            self.text = "The scraping operation timed out after 60 seconds"
        except Exception as e:
            self.title = f"Unexpected error: {str(e)}"
            self.text = f"Could not scrape with Playwright: {str(e)}"
        finally:
            # Clean up temporary file
            try:
                import os
                os.unlink(temp_script)
            except:
                pass

print("✅ Enhanced Website class loaded successfully!")
```

```python
# Smart Website Scraper with Automatic Method Detection

class SmartWebsite:
    """
    A smart website scraper that automatically detects whether to use 
    BeautifulSoup or Playwright based on the content retrieved.
    """
    
    def __init__(self, url, wait_time=2):
        self.url = url
        self.wait_time = wait_time
        self.title = ""
        self.text = ""
        self.method_used = ""
        
        # Try BeautifulSoup first (faster)
        self._try_beautifulsoup()
        
        # If BeautifulSoup didn't get good content, try Playwright
        if self._needs_javascript():
            print(f"🔄 Content seems limited with BeautifulSoup, trying Playwright for {url}")
            self._try_playwright()
    
    def _try_beautifulsoup(self):
        """Try scraping with BeautifulSoup first"""
        try:
            response = requests.get(self.url, headers=headers, timeout=10)
            soup = BeautifulSoup(response.content, 'html.parser')
            self.title = soup.title.string if soup.title else "No title found"
            
            if soup.body:
                for irrelevant in soup.body(["script", "style", "img", "input"]):
                    irrelevant.decompose()
                self.text = soup.body.get_text(separator="\n", strip=True)
            else:
                self.text = ""
                
            self.method_used = "BeautifulSoup"
            
        except Exception as e:
            self.title = f"Error: {str(e)}"
            self.text = f"Could not fetch content: {str(e)}"
            self.method_used = "BeautifulSoup (failed)"
    
    def _needs_javascript(self):
        """
        Determine if we need JavaScript rendering based on the content we got
        """
        # If we got very little content, likely needs JavaScript
        if len(self.text.strip()) < 200:
            return True
            
        # If the text contains common JavaScript framework indicators
        js_indicators = [
            "Loading...", "Please enable JavaScript", 
            "JavaScript is required", "This site requires JavaScript",
            "document.getElementById", "React", "Vue", "Angular"
        ]
        
        for indicator in js_indicators:
            if indicator in self.text:
                return True
        
        # If title suggests it's a JavaScript app
        if any(word in self.title.lower() for word in ["loading", "javascript", "enable"]):
            return True
            
        return False
    
    def _try_playwright(self):
        """Try scraping with Playwright"""
        playwright_site = EnhancedWebsite(self.url, use_playwright=True, wait_time=self.wait_time)
        
        # Only use Playwright results if they're better
        if len(playwright_site.text.strip()) > len(self.text.strip()):
            self.title = playwright_site.title
            self.text = playwright_site.text
            self.method_used = "Playwright"
        else:
            self.method_used = f"BeautifulSoup (Playwright didn't improve results)"

print("✅ Smart Website scraper loaded successfully!")
```

```python
# Website Summarizer Functions using Local Llama 3.2

# System prompt for website summarization
system_prompt = """You are an assistant that analyzes the contents of a website and provides a short summary, ignoring text that might be navigation related. Respond in markdown."""

def user_prompt_for_website(website):
    """Create a user prompt for website summarization"""
    user_prompt = f"You are looking at a website titled '{website.title}'"
    user_prompt += "\n\nThe contents of this website is as follows; please provide a short summary of this website in markdown. If it includes news or announcements, then summarize these too.\n\n"
    user_prompt += website.text
    return user_prompt

def messages_for_website(website):
    """Create messages in the format expected by OpenAI/Ollama"""
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_for_website(website)}
    ]

# Using OpenAI client format with Ollama (Method 3 from our tests)
def summarize_website_local(url, wait_time=2, model="llama3.2"):
    """
    Summarize a website using local Llama 3.2 via Ollama
    
    Args:
        url: Website URL to summarize
        wait_time: Time to wait for JavaScript (if needed)
        model: Model name to use (default: llama3.2)
    """
    print(f"🌐 Scraping website: {url}")
    
    # Use our smart scraper
    website = SmartWebsite(url, wait_time=wait_time)
    
    print(f"📖 Title: {website.title}")
    print(f"📄 Content length: {len(website.text)} characters")
    print(f"🔧 Method used: {website.method_used}")
    
    if len(website.text.strip()) < 50:
        return "⚠️ **Error**: Could not extract meaningful content from this website. The site might be blocking automated access or require special handling."
    
    print(f"🤖 Generating summary with {model}...")
    
    try:
        # Use OpenAI client format pointing to Ollama
        ollama_client = OpenAI(
            base_url="http://localhost:11434/v1",
            api_key="ollama"  # Ollama doesn't need a real API key
        )
        
        response = ollama_client.chat.completions.create(
            model=model,
            messages=messages_for_website(website)
        )
        
        summary = response.choices[0].message.content
        print("✅ Summary generated successfully!")
        return summary
        
    except Exception as e:
        print(f"❌ Error generating summary: {str(e)}")
        return f"**Error generating summary**: {str(e)}\\n\\nPlease make sure Ollama is running and the model '{model}' is available."

def display_website_summary(url, wait_time=2, model="llama3.2"):
    """Display website summary with nice markdown formatting"""
    summary = summarize_website_local(url, wait_time=wait_time, model=model)
    print("\n" + "="*60)
    print(f"SUMMARY OF: {url}")
    print("="*60)
    display(Markdown(summary))
    print("="*60)

print("✅ Website summarizer functions loaded successfully!")
```
