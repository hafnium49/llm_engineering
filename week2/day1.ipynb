## Asking LLMs to tell a joke

It turns out that LLMs don't do a great job of telling jokes! Let's compare a few models.

### What information is included in the API

Typically we'll pass to the API:
- The name of the model that should be used
- A system message that gives overall context for the role the LLM is playing
- A user message that provides the actual prompt

There are other parameters that can be used, including **temperature** which is typically between 0 and 1; higher for more random output; lower for more focused and deterministic.

````python
system_message = "You are an assistant that is great at telling jokes"
user_prompt = "Tell a light-hearted joke for an audience of Data Scientists"
````

````python
prompts = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_prompt}
  ]
````

````python
# GPT-4o-mini

completion = openai.chat.completions.create(model='gpt-4o-mini', messages=prompts)
print(completion.choices[0].message.content)
````

````python
# GPT-4.1-mini
# Temperature setting controls creativity

completion = openai.chat.completions.create(
    model='gpt-4.1-mini',
    messages=prompts,
    temperature=0.7
)
print(completion.choices[0].message.content)
````

````python
# GPT-4.1-nano - extremely fast and cheap

completion = openai.chat.completions.create(
    model='gpt-4.1-nano',
    messages=prompts
)
print(completion.choices[0].message.content)
````

````python
# GPT-4.1

completion = openai.chat.completions.create(
    model='gpt-4.1',
    messages=prompts,
    temperature=0.4
)
print(completion.choices[0].message.content)
````

````python
# If you have access to this, here is the reasoning model o4-mini
# This is trained to think through its response before replying
# So it will take longer but the answer should be more reasoned - not that this helps..

completion = openai.chat.completions.create(
    model='o4-mini',
    messages=prompts
)
print(completion.choices[0].message.content)
````

````python
# Claude 4.0 Sonnet
# API needs system message provided separately from user prompt
# Also adding max_tokens

message = claude.messages.create(
    model="claude-sonnet-4-20250514",
    max_tokens=200,
    temperature=0.7,
    system=system_message,
    messages=[
        {"role": "user", "content": user_prompt},
    ],
)

print(message.content[0].text)
````

````python
# Claude 4.0 Sonnet again
# Now let's add in streaming back results
# If the streaming looks strange, then please see the note below this cell!

result = claude.messages.stream(
    model="claude-sonnet-4-20250514",
    max_tokens=200,
    temperature=0.7,
    system=system_message,
    messages=[
        {"role": "user", "content": user_prompt},
    ],
)

with result as stream:
    for text in stream.text_stream:
            print(text, end="", flush=True)
````

## And now for some fun - an adversarial conversation between Chatbots..

You're already familar with prompts being organized into lists like:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "user prompt here"}
]
```

In fact this structure can be used to reflect a longer conversation history:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "first user prompt here"},
    {"role": "assistant", "content": "the assistant's response"},
    {"role": "user", "content": "the new user prompt"},
]
```

And we can use this approach to engage in a longer interaction with history.

````python
# Let's make a conversation between GPT-4.1-mini and Claude-3.5-haiku
# We're using cheap versions of models so the costs will be minimal

gpt_model = "gpt-4.1-mini"
claude_model = "claude-3-5-haiku-latest"

gpt_system = "You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

claude_system = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

gpt_messages = ["Hi there"]
claude_messages = ["Hi"]
````

````python
def call_gpt():
    messages = [{"role": "system", "content": gpt_system}]
    for gpt, claude in zip(gpt_messages, claude_messages):
        messages.append({"role": "assistant", "content": gpt})
        messages.append({"role": "user", "content": claude})
    completion = openai.chat.completions.create(
        model=gpt_model,
        messages=messages
    )
    return completion.choices[0].message.content
````

````python
call_gpt()
````

````python
def call_claude():
    messages = []
    for gpt, claude_message in zip(gpt_messages, claude_messages):
        messages.append({"role": "user", "content": gpt})
        messages.append({"role": "assistant", "content": claude_message})
    messages.append({"role": "user", "content": gpt_messages[-1]})
    message = claude.messages.create(
        model=claude_model,
        system=claude_system,
        messages=messages,
        max_tokens=500
    )
    return message.content[0].text
````

````python
call_claude()
````

````python
call_gpt()
````

````python
gpt_messages = ["Hi there"]
claude_messages = ["Hi"]

print(f"GPT:\n{gpt_messages[0]}\n")
print(f"Claude:\n{claude_messages[0]}\n")

for i in range(5):
    gpt_next = call_gpt()
    print(f"GPT:\n{gpt_next}\n")
    gpt_messages.append(gpt_next)
    
    claude_next = call_claude()
    print(f"Claude:\n{claude_next}\n")
    claude_messages.append(claude_next)
````

## Exercise 1: 3-Way Conversation (GPT + Claude + Gemini)

Let's create a three-way conversation between different AI models with distinct personalities!

````python
# Setup for 3-way conversation: GPT (Alex), Claude (Blake), and Gemini (Charlie)

# Models and personalities
alex_model = "gpt-4.1-mini"  # GPT as Alex
blake_model = "claude-3-5-haiku-latest"  # Claude as Blake  
charlie_model = "gemini-2.0-flash"  # Gemini as Charlie

# Define distinct personalities
alex_personality = """You are Alex, an enthusiastic and optimistic AI who loves to find the bright side of everything. 
You're energetic, use lots of exclamation points, and always try to motivate others. You're the cheerleader of the group."""

blake_personality = """You are Blake, a thoughtful and analytical AI who approaches everything with careful reasoning. 
You like to ask probing questions, consider multiple perspectives, and provide balanced, measured responses."""

charlie_personality = """You are Charlie, a creative and quirky AI with a great sense of humor. 
You love wordplay, unexpected connections, and bringing levity to serious discussions. You're the comic relief."""

# Setup Gemini client using OpenAI format for consistency
gemini_client = OpenAI(
    api_key=google_api_key, 
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

print("✅ 3-way conversation setup complete!")
print("🎭 Personalities:")
print("  🚀 Alex (GPT): Enthusiastic optimist")
print("  🤔 Blake (Claude): Thoughtful analyst") 
print("  😄 Charlie (Gemini): Creative comedian")
````

````python
# 3-Way Conversation Functions

def get_ai_response(client, model, personality, conversation_history, topic):
    """Get response from a specific AI model with personality"""
    try:
        # Build the conversation prompt
        context = f"You are {personality}\n\nConversation topic: {topic}\n\n"
        if conversation_history:
            context += "Previous conversation:\n"
            for entry in conversation_history:
                context += f"{entry['participant']}: {entry['response']}\n"
            context += "\nPlease respond to continue this conversation:"
        else:
            context += "Please start the conversation on this topic:"
        
        if 'gpt' in model:
            # OpenAI GPT
            response = client.chat.completions.create(
                model=model,
                messages=[
                    {"role": "system", "content": personality},
                    {"role": "user", "content": context}
                ],
                max_tokens=200,
                temperature=0.7
            )
            return response.choices[0].message.content.strip()
            
        elif 'claude' in model:
            # Anthropic Claude
            response = client.messages.create(
                model=model,
                max_tokens=200,
                temperature=0.7,
                messages=[
                    {"role": "user", "content": context}
                ]
            )
            return response.content[0].text.strip()
            
        elif 'gemini' in model:
            # Google Gemini (using OpenAI client format)
            response = client.chat.completions.create(
                model=model,
                messages=[
                    {"role": "system", "content": personality},
                    {"role": "user", "content": context}
                ],
                max_tokens=200,
                temperature=0.7
            )
            return response.choices[0].message.content.strip()
    
    except Exception as e:
        return f"[Error getting response: {str(e)}]"

def run_3_way_conversation(topic, rounds=3):
    """Run a 3-way conversation between Alex (GPT), Blake (Claude), and Charlie (Gemini)"""
    
    participants = [
        {
            'name': 'Alex (GPT-4o-mini)',
            'client': openai,
            'model': alex_model,
            'personality': alex_personality
        },
        {
            'name': 'Blake (Claude)',
            'client': claude,
            'model': blake_model,
            'personality': blake_personality
        },
        {
            'name': 'Charlie (Gemini)',
            'client': gemini_client,
            'model': charlie_model,
            'personality': charlie_personality
        }
    ]
    
    conversation_history = []
    
    for round_num in range(rounds):
        print(f"\n🔄 Round {round_num + 1}")
        print("-" * 60)
        
        for participant in participants:
            print(f"\n💭 {participant['name']}:")
            
            response = get_ai_response(
                participant['client'],
                participant['model'],
                participant['personality'],
                conversation_history,
                topic
            )
            
            print(f"   {response}")
            
            # Add to conversation history
            conversation_history.append({
                'participant': participant['name'],
                'response': response
            })
    
    print(f"\n✅ 3-way conversation completed with {len(conversation_history)} total responses!")
    return conversation_history

print("✅ 3-way conversation functions created!")
````

````python
# 3-Way Conversation Orchestrator

def run_three_way_conversation(starting_topic, rounds=3):
    """Run a three-way conversation between Alex, Blake, and Charlie"""
    
    print("🎬 Starting 3-way conversation!")
    print("=" * 60)
    print(f"📝 Topic: {starting_topic}")
    print("=" * 60)
    
    # Initialize conversation
    conversation = [f"Topic: {starting_topic}"]
    
    # Start with Alex
    print("\n🚀 Alex:")
    alex_response = call_alex("\\n".join(conversation))
    print(alex_response)
    conversation.append(f"Alex: {alex_response}")
    
    for round_num in range(rounds):
        print(f"\\n--- Round {round_num + 1} ---")
        
        # Blake responds
        print("\\n🤔 Blake:")
        blake_response = call_blake("\\n".join(conversation))
        print(blake_response)
        conversation.append(f"Blake: {blake_response}")
        
        # Charlie responds
        print("\\n😄 Charlie:")
        charlie_response = call_charlie("\\n".join(conversation))
        print(charlie_response)
        conversation.append(f"Charlie: {charlie_response}")
        
        # Alex responds (except on last round)
        if round_num < rounds - 1:
            print("\\n🚀 Alex:")
            alex_response = call_alex("\\n".join(conversation))
            print(alex_response)
            conversation.append(f"Alex: {alex_response}")
    
    print("\\n" + "=" * 60)
    print("🏁 Conversation complete!")
    print("=" * 60)
    
    return conversation

print("✅ 3-way conversation orchestrator ready!")
````

````python
# Let's start a 3-way conversation about AI ethics
topic = "Should AI be allowed to make decisions about human medical treatment?"

print("🗣️  Starting 3-way conversation about:", topic)
print("=" * 80)

# Run the conversation
run_3_way_conversation(topic, rounds=3)
````

## 🔄 Alternative Approach Exercise

Now let's try a different approach where all three AIs contribute to building a single comprehensive response together, rather than having separate conversations.

````python
# Alternative Approach: Collaborative Response Building
def collaborative_ai_response(topic, question):
    """
    Alternative approach where AIs build on each other's responses in sequence
    to create a more comprehensive collaborative answer
    """
    print(f"🤝 Collaborative AI Response")
    print(f"Topic: {topic}")
    print(f"Question: {question}")
    print("=" * 80)
    
    # Stage 1: Alex (GPT) provides initial analysis
    print(f"\n📊 {alex_model.upper()} Initial Analysis:")
    alex_prompt = f"""You are {alex_personality}
    
Topic: {topic}
Question: {question}

Please provide your initial analysis and thoughts. Keep it concise but insightful - other AIs will build on your response."""

    try:
        alex_response = openai.chat.completions.create(
            model=alex_model,
            messages=[{"role": "user", "content": alex_prompt}],
            max_tokens=300,
            temperature=0.7
        )
        alex_content = alex_response.choices[0].message.content.strip()
        print(alex_content)
        
    except Exception as e:
        alex_content = f"[Alex unavailable: {str(e)}]"
        print(alex_content)
    
    # Stage 2: Blake (Claude) builds on Alex's response
    print(f"\n🔍 {blake_model.upper()} Deep Dive:")
    blake_prompt = f"""You are {blake_personality}
    
Topic: {topic}
Question: {question}

Alex's initial analysis:
{alex_content}

Now build on Alex's analysis with your own deeper perspective. Consider aspects they might have missed and add analytical depth."""

    try:
        blake_response = claude.messages.create(
            model=blake_model,
            max_tokens=300,
            temperature=0.7,
            messages=[{"role": "user", "content": blake_prompt}]
        )
        blake_content = blake_response.content[0].text.strip()
        print(blake_content)
        
    except Exception as e:
        blake_content = f"[Blake unavailable: {str(e)}]"
        print(blake_content)
    
    # Stage 3: Charlie (Gemini) synthesizes and adds creative insights
    print(f"\n✨ {charlie_model.upper()} Creative Synthesis:")
    charlie_prompt = f"""You are {charlie_personality}
    
Topic: {topic}
Question: {question}

Alex's analysis:
{alex_content}

Blake's deep dive:
{blake_content}

Now synthesize their insights and add your own creative perspective. Provide a final comprehensive view that brings together all perspectives with your unique creative touch."""

    try:
        charlie_response = gemini_client.chat.completions.create(
            model=charlie_model,
            messages=[{"role": "user", "content": charlie_prompt}],
            max_tokens=350,
            temperature=0.8
        )
        charlie_content = charlie_response.choices[0].message.content.strip()
        print(charlie_content)
        
    except Exception as e:
        charlie_content = f"[Charlie unavailable: {str(e)}]"
        print(charlie_content)
    
    print(f"\n✅ Collaborative response completed!")
    
    return {
        'topic': topic,
        'question': question,
        'alex_analysis': alex_content,
        'blake_deep_dive': blake_content,
        'charlie_synthesis': charlie_content
    }

print("🤝 Alternative approach function created!")
````

## 🔄 Additional Exercise: Hybrid Cloud + Local AI Conversation

Let's enhance our 3-way conversation by replacing one of the cloud-based models with a local Ollama model. This demonstrates a **hybrid approach** combining:

- **Cloud APIs** (OpenAI GPT, Anthropic Claude) for powerful responses
- **Local Models** (Ollama) for privacy, cost control, and offline capability

We'll replace Charlie (Gemini) with a local Llama model via Ollama.

````python
# Setup Ollama Client and Hybrid Configuration
from openai import OpenAI

# Import ollama for potential direct API usage
try:
    import ollama
    print("✅ Ollama library available")
except ImportError:
    print("⚠️ Ollama library not found - will use OpenAI-compatible endpoint")

# Set up Ollama client using OpenAI-compatible endpoint
ollama_client = OpenAI(
    base_url='http://localhost:11434/v1',
    api_key='ollama'  # Ollama doesn't need a real API key
)

# Test Ollama connection
def test_ollama_connection():
    """Test if Ollama is running and has models available"""
    try:
        # Try to make a simple completion to test connection
        response = ollama_client.chat.completions.create(
            model='llama3.2',
            messages=[{"role": "user", "content": "Hi! Just testing connection."}],
            max_tokens=10
        )
        print("✅ Ollama connection successful!")
        print(f"🦙 Model responded: {response.choices[0].message.content}")
        return True
    except Exception as e:
        print(f"❌ Ollama connection failed: {str(e)}")
        print("💡 Make sure Ollama is running with: ollama serve")
        print("💡 And that llama3.2 is installed: ollama pull llama3.2")
        return False

# Hybrid model configuration (replacing Charlie with Ollama)
hybrid_alex_model = "gpt-4o-mini"
hybrid_alex_personality = "You are Alex, an enthusiastic and optimistic AI assistant. You see the bright side of technology and always focus on possibilities and opportunities. You're encouraging and forward-thinking."

hybrid_blake_model = "claude-3-haiku-20240307" 
hybrid_blake_personality = "You are Blake, a thoughtful and analytical AI assistant. You carefully consider multiple perspectives, think through implications, and provide well-reasoned responses. You're the voice of careful analysis."

# Charlie is now powered by Ollama!
hybrid_charlie_model = "llama3.2"
hybrid_charlie_personality = "You are Charlie, a creative and funny AI assistant running locally via Ollama. You bring humor, creativity, and a unique local perspective to conversations. You're witty and often think outside the box."

print("🔧 Hybrid configuration created:")
print(f"   Alex (GPT): {hybrid_alex_model}")
print(f"   Blake (Claude): {hybrid_blake_model}")  
print(f"   Charlie (Ollama): {hybrid_charlie_model}")
print()

# Test the connection
print("🧪 Testing Ollama connection...")
ollama_available = test_ollama_connection()
````

````python
# Hybrid AI Response Function (supports both cloud and local models)

def get_hybrid_ai_response(client, model, personality, conversation_history, topic):
    """Get response from AI model - works with both cloud and local models"""
    try:
        # Build the conversation prompt
        context = f"You are {personality}\n\nConversation topic: {topic}\n\n"
        if conversation_history:
            context += "Previous conversation:\n"
            for entry in conversation_history:
                context += f"{entry['participant']}: {entry['response']}\n"
            context += "\nPlease respond to continue this conversation:"
        else:
            context += "Please start the conversation on this topic:"
        
        # Handle different model types
        if 'gpt' in model or 'llama' in model:
            # OpenAI format (works for both GPT and Ollama via OpenAI-compatible API)
            response = client.chat.completions.create(
                model=model,
                messages=[
                    {"role": "system", "content": personality},
                    {"role": "user", "content": context}
                ],
                max_tokens=200,
                temperature=0.7
            )
            return response.choices[0].message.content.strip()
            
        elif 'claude' in model:
            # Anthropic Claude
            response = client.messages.create(
                model=model,
                max_tokens=200,
                temperature=0.7,
                messages=[
                    {"role": "user", "content": context}
                ]
            )
            return response.content[0].text.strip()
    
    except Exception as e:
        model_type = "Local Ollama" if 'llama' in model else "Cloud API"
        return f"[{model_type} Error: {str(e)}]"

def run_hybrid_3_way_conversation(topic, rounds=3):
    """
    Run hybrid 3-way conversation with:
    - Alex: GPT-4o-mini (Cloud)
    - Blake: Claude (Cloud) 
    - Charlie: Llama 3.2 (Local via Ollama)
    """
    
    participants = [
        {
            'name': f'Alex ({hybrid_alex_model}) ☁️',
            'client': openai,
            'model': hybrid_alex_model,
            'personality': hybrid_alex_personality
        },
        {
            'name': f'Blake ({hybrid_blake_model}) ☁️',
            'client': claude,
            'model': hybrid_blake_model,
            'personality': hybrid_blake_personality
        },
        {
            'name': f'Charlie ({hybrid_charlie_model}) 🏠',
            'client': ollama_client,
            'model': hybrid_charlie_model,
            'personality': hybrid_charlie_personality
        }
    ]
    
    print(f"🚀 HYBRID 3-WAY CONVERSATION")
    print("=" * 60)
    print(f"📝 Topic: {topic}")
    print(f"🔄 Rounds: {rounds}")
    print()
    print("🌐 Participants:")
    for p in participants:
        cloud_local = "Cloud API" if "☁️" in p['name'] else "Local Ollama"
        print(f"   {p['name']} - {cloud_local}")
    print("=" * 60)
    
    conversation_history = []
    
    for round_num in range(rounds):
        print(f"\n🔄 Round {round_num + 1}")
        print("-" * 60)
        
        for participant in participants:
            print(f"\n💭 {participant['name']}:")
            
            # Add timing for local vs cloud comparison
            import time
            start_time = time.time()
            
            response = get_hybrid_ai_response(
                participant['client'],
                participant['model'],
                participant['personality'],
                conversation_history,
                topic
            )
            
            end_time = time.time()
            response_time = end_time - start_time
            
            print(f"   {response}")
            print(f"   ⏱️ Response time: {response_time:.1f}s")
            
            # Add to conversation history
            conversation_history.append({
                'participant': participant['name'],
                'response': response
            })
    
    print(f"\n✅ Hybrid 3-way conversation completed!")
    print(f"📊 Total responses: {len(conversation_history)}")
    print("🔍 Notice the difference in response times between cloud and local models!")
    return conversation_history

print("✅ Hybrid conversation functions created!")
print("🎯 Ready to run hybrid cloud + local conversations!")
````

````python
# Test Hybrid 3-Way Conversation
# Topic: Comparing cloud vs local AI models

hybrid_topic = "What are the advantages and disadvantages of using cloud-based AI models versus local AI models?"

print("🧪 Testing Hybrid 3-Way Conversation...")
print("This will showcase the differences between:")
print("  ☁️ Cloud APIs: GPT-4o-mini & Claude (fast, powerful, internet-required)")  
print("  🏠 Local Model: Llama 3.2 via Ollama (private, offline, slower)")
print()

# Run the hybrid conversation
hybrid_conversation = run_hybrid_3_way_conversation(hybrid_topic, rounds=2)
````
